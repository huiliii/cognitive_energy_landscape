```python
# Core functionality for managing and accessing data
from neurosynth import Dataset
# Analysis tools for meta-analysis, image decoding, and coactivation analysis
from neurosynth import meta, decode, network
```

```python
dataset = Dataset.load('examples/dataset.pkl')   # Note the capital D in the second Dataset--load() is a class method
```

```python
# Run a meta-analysis on emotion
ids = dataset.get_studies('lang*', frequency_threshold=0.05)
ma = meta.MetaAnalysis(dataset, ids)
ma.save_results('.', 'language')
```

```python
# Read the CSV file and extract all words
import csv
# words = ['face recognition']
words = []
with open('ALL_word_counts_refined.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        words.append(row['Word'])

# Process each word
for word in words:
    print(f"Processing: {word}")
    try:
        ids = dataset.get_studies(word, frequency_threshold=0.05)
        ma = meta.MetaAnalysis(dataset, ids)
        ma.save_results('./data/ALL', word)
        print(f"  ✓ Completed: {word}")
    except Exception as e:
        print(f"  ✗ Error processing {word}: {e}")

print(f"\nProcessed {len(words)} words total")
```
